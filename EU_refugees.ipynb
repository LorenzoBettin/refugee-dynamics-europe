# Introduction 

In recent years, Europe has been facing a dual trend with significant humanitarian and political implications: an increasing influx of refugees (Schilling et al., 2017) and a notable shift in the political landscape towards right-wing ideologies. The rising number of refugees arriving in Europe can largely be attributed to escalating conflicts, political instability, and humanitarian crises in regions like the Middle East (Doraï, 2018), Africa, and parts of Asia. Wars, persecution, and economic collapse have forced millions of individuals and families to flee their homes in search of safety and stability (Lewtak et al., 2022), with Europe often seen as a refuge offering protection and a fresh start. Climate change, too, has compounded these pressures, leading to displacement as people in affected areas face resource scarcity, crop failure, and extreme weather events (Myers, 2002). The result is an urgent influx of refugees seeking shelter within European borders, where countries must determine how to support this vulnerable population.

At the same time, European politics has seen a marked shift towards right-wing ideologies, a trend influenced by economic anxieties, cultural identity debates, and concerns over national security. Many political parties and leaders in Europe, responding to public fears and frustrations, have adopted rhetoric and policies that focus on protecting national interests and restricting immigration (Dennison, 2019). In this context, refugees are often depicted as a burden on the economy or a threat to national identity, with right-leaning parties using them as scapegoats to address broader concerns about unemployment, housing shortages, and public safety (Anthony, 2024). This political shift can directly impact national policies, which may, in turn, determine the number of refugees each country is willing to accept and support.

This political narrative not only impacts public opinion but may also correlate with the number of refugees each country accepts or supports. To better anticipate how countries might approach housing refugees, it is essential to understand how political orientation and specific policies might affect refugee influxes. This leads us to our research questions.

## 1. CPDS dataset

### Source of the Dataset
The CPDS (Comparative Political Data Set) dataset is provided by researchers at the University of Zurich and contains political, economic, and social data covering OECD and EU member states from 1960 to 2022. The dataset is widely used in comparative political studies and includes variables on government composition, party orientation, voter turnout, economic indicators, and social policies. Its purpose is to support the empirical analysis of political and social factors across countries over time.


### Structure of the Dataset
The CPDS dataset is in Excel format and contains numerous rows representing yearly observations for each country, with over 150 columns covering various political, economic, and social variables.
This dataset contains a comprehensive time series of various socio-economic and political indicators, spanning the years 1960 to 2022 across 36 countries. The data includes 1,866 observations with 335 variables, covering attributes such as country identifiers (name, ISO codes), year, and a variety of economic, demographic, and political metrics. The dataset is downloadable as a Stata or Excel file. The file we used in this research is an Excel file.

### Variables and Data Types

The dataset includes variables like year (year of observation), country (country name), gov_right1, gov_cent1, and gov_left1 (party orientation variables indicating the proportion of right, center, and left-wing parties in government), kaopen (capital account openness), debt (government debt), vturn (voter turnout), pop (population), and various economic indicators such as inflation and realgdpgr (real GDP growth). These variables are a blend of numerical and categorical data, providing a robust view of each country's political, social, and economic landscape.

Due to the dataset's extensive scope, with 335 variables covering socio-economic, political, and demographic metrics, listing each individually would be impractical and would hinder readability. Instead, focusing on the main categories—like government composition, economic conditions, social policies, and demographic information—provides a concise yet comprehensive overview of the dataset’s structure and core areas of interest. This summarization helps to highlight the dataset’s capacity for enabling detailed, multidimensional analysis of political and economic trends over time.
The dataset includes essential identifiers like year (integer), country (categorical), iso (categorical), and iso3n (integer), which define each entry by country and year. Demographic indicators, such as pop (float, total population), pop15_64 (float, working-age population), and pop65 (float, elderly population), provide insight into population distributions by age. Political metrics, like gov_right1 (float, proportion of right-leaning government), capture political affiliations. These key variables span primarily numerical types (integer and float) for continuous data, with categorical fields for country identifiers and political affiliations, providing a well-rounded framework for analyzing long-term socio-economic and political trends. 
A complete overview of this dataset can be found at <https://cpds-data.org/wp-content/uploads/2024/09/codebook_cpds-.pdf>

```python
import pandas as pd
import numpy as np
from sklearn.impute import SimpleImputer
import statsmodels.api as sm
from sklearn.utils import resample
from pathlib import Path

# Step 1: Define the data directory
DATA_DIR = Path("data")

# Step 2: Load the datasets
df1 = pd.read_excel(
    DATA_DIR / "cpds-1960-2022-update-2024 (3).xlsx"
)

df2 = pd.read_csv(
    DATA_DIR / "API_SM.POP.REFG_DS2_en_csv_v2_10798 (3).csv",
    skiprows=4
)
```

# Preprocessing 


In the code cell below the preprocessing is shown. This cell will describe the steps that have been taken to preprocess this data and explain why each step has been taken.

Step 1 Reshaping and Filtering the Refugee Data: The initial step reshapes the refugee dataset by melting it, transforming year columns into rows under a year column. This restructuring makes the data compatible for time-series analysis, where each row represents a specific year for each country. The data is then filtered to include only numeric years, converting the year column to integers to standardize the data type.

Step 2 Merging Datasets: After reshaping, the refugee dataset is merged with the main dataset on Country Code and year. The merge uses an inner join to ensure only data points with corresponding country codes and years in both datasets are included. Rows with missing values in refugee_population are removed to ensure a clean target variable for analysis.

Step 3 Defining Core Columns: Numerical and categorical columns essential for the analysis are defined based on the core indicators and the dataset's codebook. This helps streamline the dataset to relevant features, improving efficiency for downstream modeling tasks.

Step 4 Encoding Categorical Data: Categorical columns are dummy encoded, transforming categories into binary columns that can be used in numerical models. This step enables seamless integration of categorical data into the analysis, ensuring all variables are compatible with machine learning algorithms.

Step 5 Imputing Missing Values: Finally, any remaining missing values in the selected features are imputed using the mean strategy. This ensures that no missing data interferes with the analysis, resulting in a fully prepared and clean dataset for modeling. 

```python
# Step 2: Reshape and filter the refugee data
df2_melted = df2.melt(id_vars=["Country Name", "Country Code", "Indicator Name", "Indicator Code"], 
                      var_name="year", value_name="refugee_population")
df2_melted_clean = df2_melted[df2_melted['year'].apply(lambda x: x.isnumeric())]
df2_melted_clean['year'] = df2_melted_clean['year'].astype(int)
df2_filtered = df2_melted_clean[df2_melted_clean['Country Code'].isin(df1['iso'])]

# Step 3: Merge the datasets and filter out rows with missing target variable
merged_df = pd.merge(df1, df2_filtered, left_on=["iso", "year"], right_on=["Country Code", "year"], how="inner")
merged_df = merged_df.dropna(subset=['refugee_population'])

# Step 4: Define the core numerical and categorical columns to include, adding additional ones based on the codebook
core_numerical_columns = [ 
     'dir', 'openc', 'kaopen', 'receipts', 'inflation', 
     'debt', 'ttl_labf', 'pop', 'unemp', 'vturn', 
     'educatt_minimal', 'womanpar', 'realgdpgr', 'urbanpop', 
     'postfisc_gini', 'leff'
]

categorical_columns = ['gov_party', 'poco', 'eu', 'gov_type', 'fed', 'pres', 'prop', 'bic', 'referen']

# Filter columns based on their availability in the dataset
available_columns = merged_df.columns.tolist()
numerical_columns = [col for col in core_numerical_columns if col in available_columns]
categorical_columns = [col for col in categorical_columns if col in available_columns]

# Step 5: Dummy encode categorical columns and add them to the numerical columns
merged_df_dummies = pd.get_dummies(merged_df, columns=categorical_columns, drop_first=True)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Reset index if needed
average_df = average_df.reset_index(drop=True)

# Define Schmidt-Index color palette and labels
schmidt_index_labels = {
    1: "Hegemony of right-wing (and center) parties",
    2: "Dominance of right-wing (and center) parties",
    3: "Balance of power between left and right",
    4: "Dominance of social-democratic and other left parties",
    5: "Hegemony of social-democratic and other left parties"
}
palette = sns.color_palette("viridis", 5)  # Create a color palette for 5 distinct levels
color_map = dict(zip(schmidt_index_labels.keys(), palette))  # Map levels to colors

# Assign Schmidt-Index based on intervals
def assign_schmidt_index(value):
    if value == 1:
        return 1
    elif 1 < value <= 2.33:
        return 2
    elif 2.33 < value < 3.66:
        return 3
    elif 3.66 <= value < 5:
        return 4
    elif value == 5:
        return 5

# Apply the function to create a new column for color-coded Schmidt-Index levels
average_df['schmidt_index'] = average_df['gov_party'].apply(assign_schmidt_index)

# Create figure and scatter plot
plt.figure(figsize=(12, 8))

# Scatter plot with color-coding based on Schmidt-Index
for index in schmidt_index_labels:
    subset = average_df[average_df['schmidt_index'] == index]
    plt.scatter(subset['Population'], subset['refugee_population'], 
                color=color_map[index], label=schmidt_index_labels[index], alpha=0.7, s=50)

# Set log scale if population values vary widely
plt.xscale('log')
plt.yscale('log')

# Add text labels with increased font size
for i in range(len(average_df)):
    plt.text(average_df.iloc[i]['Population'], average_df.iloc[i]['refugee_population'], 
             average_df.iloc[i]['country'], fontsize=9, ha='right', va='bottom')

# Adjust axes limits to focus on clustered data points (modify values as needed)
plt.xlim(3e5, 3e8)  # Adjusted example limits for population
plt.ylim(1e3, 7e5)  # Adjusted example limits for refugee population

# Labels and title
plt.xlabel("General Population (log scale)", fontsize=14)
plt.ylabel("Average Refugee Population (log scale)", fontsize=14)
plt.title("Average Refugee Population vs General Population by Country", fontsize=16)

# Grid for better readability
plt.grid(True, which="both", linestyle='--', linewidth=0.5)

# Add legend for Schmidt-Index color coding
plt.legend(title='Cabinet Composition (Schmidt-Index)')

# Show plot
plt.show()
```

The interactive diagram below provides a visual representation of refugee population trends over time, allowing the exploration of data by selecting a specific country or viewing global totals.
```python
# Create dropdown for country selection, sliders for year range, and option for Total totals
from ipywidgets import interact
import ipywidgets as widgets

countries = list(merged_df['country'].unique())
countries.insert(0, 'Total')  # Add "Total" option to the top
years = merged_df['year'].unique()

def plot_refugee_population(country, start_year, end_year):
    plt.figure(figsize=(10, 6))

    if country == 'Total':
        # Aggregate Total refugee population per year
        Total_data = merged_df[(merged_df['year'] >= start_year) & (merged_df['year'] <= end_year)]
        Total_totals = Total_data.groupby('year')['refugee_population'].sum().reset_index()
        plt.plot(Total_totals['year'], Total_totals['refugee_population'], marker='o', color='r', label='Total Refugee Population')
        plt.title(f"Total Refugee Population from {start_year} to {end_year}")
    else:
        # Filter data for the selected country and year range
        country_data = merged_df[(merged_df['country'] == country) & 
                                 (merged_df['year'] >= start_year) & 
                                 (merged_df['year'] <= end_year)]
        plt.plot(country_data['year'], country_data['refugee_population'], marker='o', color='b', label=f'Refugee Population in {country}')
        plt.title(f"Refugee Population in {country} from {start_year} to {end_year}")

    plt.xlabel("Year")
    plt.ylabel("Refugee Population")
    plt.legend()
    plt.grid(True)
    plt.show()

# Interactive widgets for country selection, start and end year sliders
interact(
    plot_refugee_population,
    country=widgets.Dropdown(options=countries, description='Country'),
    start_year=widgets.IntSlider(value=min(years), min=min(years), max=max(years)-1, step=1, description='Start Year'),
    end_year=widgets.IntSlider(value=max(years), min=min(years), max=max(years), step=1, description='End Year')
)
```

# Machine learning 

This code follows a comprehensive approach to preparing, training, evaluating, and cross-validating a linear regression model to predict refugee populations. It starts by preparing the dataset, filtering columns to include only relevant numerical and categorical variables. Categorical variables are converted to a numeric format through dummy encoding, which is essential for integrating them effectively into the regression model. Missing values in the dataset are imputed using the mean strategy, ensuring that all features are complete and usable for modeling, which minimizes data loss and enhances model reliability.

Once data preprocessing is complete, the dataset is split into training and test sets to allow for independent evaluation of the model’s predictive accuracy. A constant term is added to the model to capture baseline effects, and an Ordinary Least Squares (OLS) regression is fitted using the training data. After training, the model’s performance is evaluated on the test set using Mean Squared Error (MSE) and R-squared metrics, which provide insights into prediction accuracy and explained variance. To further assess model generalizability, cross-validation is performed using R-squared scores across multiple folds, with an average score calculated to summarize the model’s predictive stability across different data subsets. This structured approach ensures the model is both robustly trained and rigorously validated, balancing accuracy with the prevention of overfitting.

```python
from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.metrics import mean_squared_error, r2_score
from sklearn.linear_model import LinearRegression

# Full Dataset

# Filter columns based on their availability in the dataset
available_columns = merged_df.columns.tolist()
numerical_columns = [col for col in core_numerical_columns if col in available_columns]
categorical_columns = [col for col in categorical_columns if col in available_columns]

# Step 5: Dummy encode categorical columns and add them to the numerical columns
merged_df_dummies = pd.get_dummies(merged_df, columns=categorical_columns, drop_first=True)
X = merged_df_dummies[numerical_columns + [col for col in merged_df_dummies.columns if col.startswith(tuple(categorical_columns))]]
y = merged_df['refugee_population']

# Step 6: Impute missing values in X
imputer = SimpleImputer(strategy='mean')
X_imputed = imputer.fit_transform(X)
X_imputed_df = pd.DataFrame(X_imputed, columns=X.columns).reset_index(drop=True)
y = y.reset_index(drop=True)

# Step 7 :Split the data
X_train, X_test, y_train, y_test = train_test_split(
    X_imputed_df, y, test_size=0.2, random_state=42
)

# Step 8 :Add a constant
X_train_const = sm.add_constant(X_train)
X_test_const = sm.add_constant(X_test)

# Step 9 :Fit the model
model = sm.OLS(y_train, X_train_const)
results = model.fit()

# Print the model summary
print("\nModel Summary with All Variables:")
print(results.summary())

# Step 10 :Evaluate the model
y_pred = results.predict(X_test_const)
mse = mean_squared_error(y_test, y_pred)
r_squared = r2_score(y_test, y_pred)

print("\nModel Evaluation on Test Set:")
print(f"Mean Squared Error: {mse}")
print(f"R-squared: {r_squared}")

# Step 11 :Cross-validation
lr = LinearRegression()
cv_scores = cross_val_score(lr, X_imputed_df, y, cv=5, scoring='r2')

print("\nCross-Validation R-squared Scores:")
print(cv_scores)
print(f"Average R-squared Score: {cv_scores.mean()}")
```

A thorough preliminary model, grounded in research and a clear understanding of the variables involved, provides a foundation that streamlines the coding phase, reducing the likelihood of costly revisions and enabling a more focused, accurate analysis.
